Q7.1. Write a function my_abs(val) that returns the absolute value of value without using Python's built-in abs().
Test it with a positive number, a negative number, and 0.

```python
def my_abs(val):
    if val < 0:
        return -val
    else:
        return val

print(my_abs(10))
print(my_abs(-10))
print(my_abs(0))
```

Q7.3. Write a function print_abs(val) that prints (not returns) the absolute value. Then run x = print_abs(-2.7);
print(x). What gets printed for x, and why?

Ans:-x is None because print_abs() prints the value but does not use return. A function that doesn't return a value automatically returns None.

Q11. Class Attendance Counter

• Create a global variable total_students = 0.

to use the global keyword inside the function so it can change the global variable.

• Call mark_present() three times (once for each student who walks in), then print

Expected output:

```python
total_students = 0

def mark_present():
    global total_students
    total_students += 1

mark_present()
mark_present()
mark_present()

print(total_students)
```
